```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("results/3-baidu-ultr/test.csv")
df.head()
```

```python
base = alt.Chart(df, width=500)

bars = base.mark_bar().encode(
    x=alt.X("model", title="", sort="y"),
    y=alt.Y("mean(test_ppl)").scale(zero=False, domain=(1.3, 1.42))
)

errors = base.mark_errorbar(extent="ci", thickness=4).encode(
    x=alt.X("model", title=""),
    y=alt.Y("test_ppl").scale(zero=False, domain=(1.3, 1.42))
)

bars + errors
bars
```

```python
base = alt.Chart(df, width=500)

bars = base.mark_bar().encode(
    x=alt.X("model", title="", sort="y"),
    y=alt.Y("mean(test_cond_ppl)").scale(zero=False, domain=(1.28, 1.42), clamp=True)
)

errors = base.mark_errorbar(extent="ci", thickness=4).encode(
    x=alt.X("model", title=""),
    y=alt.Y("test_cond_ppl").scale(zero=False, domain=(1.15, 1.42))
)

bars + errors
bars
```
